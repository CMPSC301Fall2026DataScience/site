# Data Analysis with R

This notebook demonstrates data analysis and visualization techniques using R in JupyterLite.

## Creating Sample Data

Let's create a dataset to analyze.

```python
# Set seed for reproducibility
set.seed(123)

# Create a sample dataset
n <- 100
grades <- data.frame(
  student_id = 1:n,
  study_hours = round(rnorm(n, mean = 5, sd = 2), 1),
  attendance = round(runif(n, min = 60, max = 100)),
  previous_gpa = round(rnorm(n, mean = 3.0, sd = 0.5), 2)
)

# Calculate final grade based on factors
grades$final_grade <- round(
  60 + 
  (grades$study_hours * 3) + 
  (grades$attendance * 0.2) + 
  (grades$previous_gpa * 5) + 
  rnorm(n, mean = 0, sd = 5)
)

# Ensure grades are between 0 and 100
grades$final_grade <- pmax(0, pmin(100, grades$final_grade))

# Show first few rows
head(grades, 10)
```

## Summary Statistics

Let's examine the distribution of our variables.

```python
# Overall summary
summary(grades)

# Calculate custom statistics
cat("\nAdditional Statistics:\n")
cat("Mean Final Grade:", mean(grades$final_grade), "\n")
cat("Median Final Grade:", median(grades$final_grade), "\n")
cat("SD Final Grade:", sd(grades$final_grade), "\n")
cat("\nPercentiles:\n")
quantile(grades$final_grade, probs = c(0.25, 0.5, 0.75, 0.95))
```

## Correlation Analysis

Examine relationships between variables.

```python
# Calculate correlation matrix
cor_matrix <- cor(grades[, c("study_hours", "attendance", "previous_gpa", "final_grade")])
print(round(cor_matrix, 3))

# Test correlation significance
cat("\nCorrelation between Study Hours and Final Grade:\n")
cor.test(grades$study_hours, grades$final_grade)
```

## Linear Regression

Build a model to predict final grades.

```python
# Build linear regression model
model <- lm(final_grade ~ study_hours + attendance + previous_gpa, data = grades)

# Display model summary
summary(model)

# Extract coefficients
cat("\nModel Coefficients:\n")
print(coef(model))
```

## Making Predictions

Use the model to predict grades for new students.

```python
# Create new student data
new_students <- data.frame(
  study_hours = c(3, 6, 8),
  attendance = c(70, 85, 95),
  previous_gpa = c(2.5, 3.2, 3.8)
)

# Make predictions
predictions <- predict(model, newdata = new_students, interval = "confidence")

# Display results
result <- cbind(new_students, round(predictions, 1))
print(result)
```

## Data Visualization

Create visualizations to understand the data better.

```python
# Histogram of final grades
hist(grades$final_grade,
     breaks = 20,
     col = "skyblue",
     border = "white",
     main = "Distribution of Final Grades",
     xlab = "Final Grade",
     ylab = "Frequency")
abline(v = mean(grades$final_grade), col = "red", lwd = 2, lty = 2)
legend("topright", legend = "Mean", col = "red", lwd = 2, lty = 2)
```

```python
# Scatter plot with regression line
plot(grades$study_hours, grades$final_grade,
     pch = 19,
     col = rgb(0, 0, 1, 0.5),
     main = "Study Hours vs Final Grade",
     xlab = "Study Hours per Week",
     ylab = "Final Grade")
abline(lm(final_grade ~ study_hours, data = grades), 
       col = "red", lwd = 2)
grid()
```

```python
# Create grade categories
grades$letter_grade <- cut(grades$final_grade,
                           breaks = c(0, 60, 70, 80, 90, 100),
                           labels = c("F", "D", "C", "B", "A"),
                           include.lowest = TRUE)

# Boxplot by grade category
boxplot(study_hours ~ letter_grade, data = grades,
        col = c("red", "orange", "yellow", "lightgreen", "green"),
        main = "Study Hours by Letter Grade",
        xlab = "Letter Grade",
        ylab = "Study Hours per Week")
```

## Grouped Analysis

Analyze data by groups.

```python
# Summary by letter grade
aggregate_result <- aggregate(cbind(study_hours, attendance, previous_gpa) ~ letter_grade,
                              data = grades,
                              FUN = mean)

print("Average values by letter grade:")
print(round(aggregate_result, 2))

# Count by letter grade
cat("\nGrade distribution:\n")
print(table(grades$letter_grade))
```
